```python

import re

with open("text.txt", "r", encoding="utf-8") as file:
    text = file.read()

print("Текст успішно прочитано.")
print("Початок тексту:")
print(text[:300])
```

```python

# Функція для отримання слів із тексту
def get_words(text):
    words = re.findall(r"[а-щьюяґєіїА-ЩЬЮЯҐЄІЇ]+(?:['’ʼ-][а-щьюяґєіїА-ЩЬЮЯҐЄІЇ]+)*", text.lower())
    return words


# Частотний словник у відсотках
def word_frequency(text):
    words = get_words(text)
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    for word in frequency:
        frequency[word] = round(frequency[word] / len(words) * 100, 2)

    return frequency


# Найкоротші слова
def shortest_words(text):
    words = get_words(text)
    min_length = min(len(word) for word in words)
    return sorted(set(word for word in words if len(word) == min_length))


# Найдовші слова
def longest_words(text):
    words = get_words(text)
    max_length = max(len(word) for word in words)
    return sorted(set(word for word in words if len(word) == max_length))


# Кількість унікальних слів
def unique_words_count(text):
    words = get_words(text)
    return len(set(words))


# Загальна кількість слів
def words_count(text):
    words = get_words(text)
    return len(words)


# Літери, які використовуються в тексті
def used_letters(text):
    letters = {}

    for letter in text.lower():
        if letter.isalpha():
            if letter in letters:
                letters[letter] += 1
            else:
                letters[letter] = 1

    return letters


# Найчастіша літера
def most_common_letter(text):
    letters = used_letters(text)
    max_count = max(letters.values())

    for letter in letters:
        if letters[letter] == max_count:
            return letter
```

```python

print("АНАЛІЗ ТЕКСТУ")
print()

print("Кількість слів:", words_count(text))
print("Кількість унікальних слів:", unique_words_count(text))
print("Найкоротші слова:", shortest_words(text))
print("Найдовші слова:", longest_words(text))

print()
print("Літери та їх кількість:")
print(used_letters(text))

print()
print("Найчастіша літера:", most_common_letter(text))

print()
print("10 найчастіших слів:")

frequency = word_frequency(text)
sorted_words = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

for word, percent in sorted_words[:10]:
    print(word, "-", percent, "%")
```
